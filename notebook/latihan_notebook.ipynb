```python
# Bandingkan pasokan uap rencana vs uap aktual per unit.
import sys; sys.path.insert(0, "..")
import pandas as pd
import matplotlib.pyplot as plt
import gaya_grafik as gg

produksi = pd.read_csv("../data/produksi_bulanan.csv")

gg.terapkan()
fig, ax = plt.subplots()

for unit in ["ARL", "BTL", "CKR", "DRG", "EKL", "FRS"]:
    sub = produksi[produksi["kode_unit"].str.upper() == unit]
    ax.scatter(
        sub["uap_rencana_ton"],
        sub["uap_aktual_ton"],
        color=gg.warna_unit(unit),
        label=unit,
        alpha=0.6,
        s=25,
    )

batas = max(produksi["uap_rencana_ton"].max(), produksi["uap_aktual_ton"].max())
ax.plot([0, batas], [0, batas], "--", color=gg.ABU, linewidth=1.2, label="Target 1:1")

ax.legend(ncol=4, loc="upper left")
gg.rapikan(ax, judul="Uap Rencana vs Uap Aktual per Unit", x="Uap Rencana (ton)", y="Uap Aktual (ton)")
plt.show()
```
