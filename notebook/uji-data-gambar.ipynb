```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

DATA = Path("..") / "data"
```

## 1. Citra termal

```python
label = pd.read_csv(DATA / "citra_termal" / "label.csv")
print(len(label), "baris")          # 160
label["label"].value_counts()       # normal 80, titik_panas 80
```

```python
# satu gambar = tabel angka: tinggi x lebar x 3 warna (R, G, B)
gambar = np.asarray(Image.open(DATA / "citra_termal" / "T001.png"))
print(gambar.shape, gambar.dtype)   # (120, 160, 3) uint8
```

```python
# empat contoh dari tiap label
fig, ax = plt.subplots(2, 4, figsize=(12, 5))
for baris, lab in enumerate(["normal", "titik_panas"]):
    contoh = label[label["label"] == lab].head(4)
    for kolom, (_, r) in enumerate(contoh.iterrows()):
        ax[baris, kolom].imshow(Image.open(DATA / "citra_termal" / r["file"]))
        ax[baris, kolom].set_title(f"{r['file']} · {lab} · {r['suhu_maks_c']} °C", fontsize=9)
        ax[baris, kolom].axis("off")
plt.tight_layout()
plt.show()
```

## 2. Formulir scan
